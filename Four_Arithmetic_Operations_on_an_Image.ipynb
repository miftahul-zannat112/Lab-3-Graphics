```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
uploaded = files.upload()

image_names = list(uploaded.keys())

if len(image_names) < 2:
    raise ValueError("Please upload two images!")
```

```python
img1 = cv2.imread(image_names[0])
img2 = cv2.imread(image_names[1])

img2 = cv2.resize(
    img2,
    (img1.shape[1], img1.shape[0])
)
```

```python
addition = cv2.add(img1, img2)
```

```python
subtraction = cv2.subtract(img1, img2)
```

```python
multiplication = cv2.multiply(
    img1,
    img2,
    scale=1/255.0
)
```

```python
division = cv2.divide(
    img1,
    np.maximum(img2, 1)
)
```

```python
np.maximum(img2, 1)
```

```python
img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2 = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)

addition = cv2.cvtColor(
    addition, cv2.COLOR_BGR2RGB
)

subtraction = cv2.cvtColor(
    subtraction, cv2.COLOR_BGR2RGB
)

multiplication = cv2.cvtColor(
    multiplication, cv2.COLOR_BGR2RGB
)

division = cv2.cvtColor(
    division, cv2.COLOR_BGR2RGB
)
```

```python
plt.figure(figsize=(14, 9))

plt.subplot(2, 3, 1)
plt.imshow(img1)
plt.title("Input Image 1")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(img2)
plt.title("Input Image 2")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(addition)
plt.title("Addition")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(subtraction)
plt.title("Subtraction")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(multiplication)
plt.title("Multiplication")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(division)
plt.title("Division")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
cv2.imwrite(
    "Addition_Output.png",
    cv2.cvtColor(addition, cv2.COLOR_RGB2BGR)
)

cv2.imwrite(
    "Subtraction_Output.png",
    cv2.cvtColor(subtraction, cv2.COLOR_RGB2BGR)
)

cv2.imwrite(
    "Multiplication_Output.png",
    cv2.cvtColor(multiplication, cv2.COLOR_RGB2BGR)
)

cv2.imwrite(
    "Division_Output.png",
    cv2.cvtColor(division, cv2.COLOR_RGB2BGR)
)

print("All Four Arithmetic Operations Completed Successfully!")
```
